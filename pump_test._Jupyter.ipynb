```python
import logging
import pycont.controller
from pycont.controller import PumpIO


logging.basicConfig(level=logging.INFO)

SETUP_CONFIG_FILE = 'C:/Users/Blab2/Desktop/pycont/pump_setup_config.json'
controller = pycont.controller.MultiPumpController.from_configfile(SETUP_CONFIG_FILE)

# Initialize the controller
pump = controller.pumps['water']
pump.flash_eeprom_3_way_y_valve()  # sets valve type in EEPROM
pump.initialize()                   # hard reset to volume 0
```

```python
pump.smart_initialize()
pump.go_to_volume(5, wait=True)  # go to volume 0
```
